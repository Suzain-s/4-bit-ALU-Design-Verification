```python
def alu(a,b,operation):

    if operation=="ADD":
      return(a+b) & 0b1111

    elif operation=="SUB":
      return(a-b) & 0b1111

    elif operation=="AND":
      return(a&b)

    elif operation=="OR":
      return(a|b)

    elif operation=="XOR":
       return(a^b)

    else:
      raise ValueError("Invalid operation")

def verify_alu(a,b,operation,expected):

  actual = alu(a,b,operation)

  if actual == expected:
    print("PASS")
  else:
    print("FAIL")

  print("A:", a)
  print("B:", b)
  print("Operation:", operation)
  print("Expected:", expected)
  print("Actual:", actual)
  print("--------------------")

verify_alu(3,2,"ADD",5)
verify_alu(7,3,"SUB",4)
verify_alu(12,10,"AND",8)
verify_alu(12,10,"OR",14)
verify_alu(12,10,"XOR",6)




```

```python
operations=["ADD","SUB","AND","OR","XOR"]

total_tests=0
passed=0
failed=0

for operation in operations:
  for a in range(16):
    for b in range (16):

      if operation == "ADD":
        expected = (a+b) & 0b1111

      elif operation == "SUB":
        expected = (a-b) & 0b1111

      elif operation == "AND":
        expected = (a&b)

      elif operation == "OR":
        expected = (a|b)

      elif operation == "XOR":
        expected = (a^b)

      actual = alu(a,b,operation)

      total_tests += 1

      if actual == expected:
        passed += 1
      else:
        failed += 1
        print("FAIL:",a,b,operation)
        print("Expected:", expected,"Actual:", actual)
print("\n===== FINAL REPORT =====")
print("Total tests:", total_tests)
print("Passed:",passed)
print("Failed:",failed)

print("\n===== BOUNDARY TEST =====")

verify_alu(15,1,"ADD",0)
verify_alu(0,1,"SUB",15)
verify_alu(15,15,"AND",15)
verify_alu(0,0,"OR",0)
verify_alu(15,0,"XOR",15)
```
